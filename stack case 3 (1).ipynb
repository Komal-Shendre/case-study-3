```python
# stack
# Implementation of Stack and Queue Using Linked Lists

# Node class
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# Stack implementation using Linked List
class Stack:
    def __init__(self):
        self.top = None

    # Insert element into stack
    def push(self, data):
        new_node = Node(data)
        new_node.next = self.top
        self.top = new_node
        print(data, "pushed into Stack.")

    # Remove element from stack
    def pop(self):
        if self.top is None:
            print("Stack is empty!")
            return

        data = self.top.data
        self.top = self.top.next
        print("Popped element:", data)

    # Display top element
    def peek(self):
        if self.top is None:
            print("Stack is empty!")
        else:
            print("Top element:", self.top.data)

    # Display all stack elements
    def display(self):
        if self.top is None:
            print("Stack is empty!")
            return

        current = self.top
        print("Stack elements (Top to Bottom):", end=" ")

        while current:
            print(current.data, end=" ")
            current = current.next

        print()


# Queue implementation using Linked List
class Queue:
    def __init__(self):
        self.front = None
        self.rear = None

    # Insert element into queue
    def enqueue(self, data):
        new_node = Node(data)

        if self.rear is None:
            self.front = self.rear = new_node
        else:
            self.rear.next = new_node
            self.rear = new_node

        print(data, "inserted into Queue.")

    # Remove element from queue
    def dequeue(self):
        if self.front is None:
            print("Queue is empty!")
            return2

        data = self.front.data
        self.front = self.front.next

        if self.front is None:
            self.rear = None

        print("Deleted element:", data)

    # Display front element
    def show_front(self):
        if self.front is None:
            print("Queue is empty!")
        else:
            print("Front element:", self.front.data)

    # Display all queue elements
    def display(self):
        if self.front is None:
            print("Queue is empty!")
            return

        current = self.front
        print("Queue elements (Front to Rear):", end=" ")

        while current:
            print(current.data, end=" ")
            current = current.next

        print()


# Main program
stack = Stack()
queue = Queue()

while True:
    print("\n===== STACK AND QUEUE USING LINKED LIST =====")
    print("1. Stack Operations")
    print("2. Queue Operations")
    print("3. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        while True:
            print("\n--- STACK MENU ---")
            print("1. Push")
            print("2. Pop")
            print("3. Peek")
            print("4. Display Stack")
            print("5. Back to Main Menu")

            option = input("Enter your choice: ")

            if option == "1":
                try:
                    value = int(input("Enter element: "))
                    stack.push(value)
                except ValueError:
                    print("Please enter a valid integer.")

            elif option == "2":
                stack.pop()

            elif option == "3":
                stack.peek()

            elif option == "4":
                stack.display()

            elif option == "5":
                break

            else:
                print("Invalid choice!")

    elif choice == "2":
        while True:
            print("\n--- QUEUE MENU ---")
            print("1. Enqueue")
            print("2. Dequeue")
            print("3. Front")
            print("4. Display Queue")
            print("5. Back to Main Menu")

            option = input("Enter your choice: ")

            if option == "1":
                try:
                    value = int(input("Enter element: "))
                    queue.enqueue(value)
                except ValueError:
                    print("Please enter a valid integer.")

            elif option == "2":
                queue.dequeue()

            elif option == "3":
                queue.show_front()

            elif option == "4":
                queue.display()

            elif option == "5":
                break

            else:
                print("Invalid choice!")

    elif choice == "3":
        print("Program terminated successfully.")
        break

    else:
        print("Invalid choice! Please try again.")
```
